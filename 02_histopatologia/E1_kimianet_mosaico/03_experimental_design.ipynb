## 1. Configuration

```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("data")     # adjust to the location of the input files

GENE_EXPRESSION_FILE = DATA_DIR / "gene_expression.tsv"
SUBTYPE_FILE = DATA_DIR / "subtype.tsv"
PAIRED_PATIENTS_FILE = DATA_DIR / "pacientes_emparejados_1081.tsv"
```

## 2. Loading the tables

```python
gene_expression = pd.read_csv(GENE_EXPRESSION_FILE, sep="\t", low_memory=False)
subtype_table = pd.read_csv(SUBTYPE_FILE, sep="\t", low_memory=False)
paired_patients = pd.read_csv(PAIRED_PATIENTS_FILE, sep="\t", low_memory=False)

for name, table in [("gene_expression", gene_expression), ("subtype_table", subtype_table),
                    ("paired_patients", paired_patients)]:
    print(f"{name}: {table.shape[0]:,} rows x {table.shape[1]:,} columns")
```

## 4. Subtype matrix

`subtype.tsv` stores the expression data as genes × patients, with a `Subtype` row. The matrix is transposed to a patients × genes layout, in which the patient identifier is the index and the subtype is a regular column.

```python
subtype_by_patient = subtype_table.T
subtype_by_patient.columns = subtype_table["gene_name"]
subtype_by_patient = subtype_by_patient.iloc[1:].copy()        # drop the row holding the 'gene_name' column
subtype_by_patient["ID"] = subtype_by_patient.index

print(f"Patients x columns: {subtype_by_patient.shape}")
print("\nPatients per subtype:")
print(subtype_by_patient["Subtype"].value_counts().to_string())
subtype_by_patient[["ID", "Subtype"]].head()
```
